# Module 5: Route Efficiency Benchmarking

```python
import pandas as pd
import numpy as np
```

```python
# Load Feature Engineered Dataset

df = pd.read_csv("../data/processed/feature_engineered_dataset.csv")
```

```python
df.head()
```

```python
# Route Summary
route_summary = (
    df.groupby("Factory_to_State")
      .agg(
          Shipment_Count=("Shipping Lead Time", "count"),
          Average_Lead_Time=("Shipping Lead Time", "mean"),
          Minimum_Lead_Time=("Shipping Lead Time", "min"),
          Maximum_Lead_Time=("Shipping Lead Time", "max")
      )
      .reset_index()
)

route_summary.head()
```

```python
# Fastest Routes
fastest_routes = route_summary.sort_values(
    by="Average_Lead_Time",
    ascending=True
)

fastest_routes.head(10)
```

```python
# Slowest Routes
slowest_routes = route_summary.sort_values(
    by="Average_Lead_Time",
    ascending=False
)

slowest_routes.head(10)
```

```python
# Route Ranking
route_summary["Route_Rank"] = (
    route_summary["Average_Lead_Time"]
    .rank(method="dense", ascending=True)
)

route_summary.head()
```

```python
# Top 10 Efficient Routes
top10_fastest = fastest_routes.head(10)

top10_fastest
```

```python
# Bottom 10 Efficient Routes
bottom10_routes = slowest_routes.head(10)

bottom10_routes
```

```python
# Overall Statistics
route_summary.describe()
```

```python
top10_fastest = fastest_routes.head(10)
top10_slowest = slowest_routes.head(10)
```

```python
# Save Results

route_summary.to_csv(
    "../data/processed/route_efficiency_benchmark.csv",
    index=False
)

top10_fastest.to_csv(
    "../data/processed/top10_fastest_routes.csv",
    index=False
)

top10_slowest.to_csv(
    "../data/processed/top10_slowest_routes.csv",
    index=False
)
```
