# Module 8 : KPI Calculation and Business Insights

```python
import pandas as pd
import plotly.express as px
```

```python
# Load Dataset

df = pd.read_csv("../data/processed/feature_engineered_dataset.csv")
```

```python
df.head()
```

```python
# Overall KPIs
total_shipments = len(df)
total_sales = df["Sales"].sum()
total_profit = df["Gross Profit"].sum()
average_lead_time = df["Shipping Lead Time"].mean()

print("Total Shipments :", total_shipments)
print("Total Sales :", round(total_sales,2))
print("Total Profit :", round(total_profit,2))
print("Average Lead Time :", round(average_lead_time,2))
```

```python
# KPI DataFrame
kpi_summary = pd.DataFrame({
    "KPI":[
        "Total Shipments",
        "Total Sales",
        "Total Profit",
        "Average Lead Time"
    ],
    "Value":[
        total_shipments,
        round(total_sales,2),
        round(total_profit,2),
        round(average_lead_time,2)
    ]
})

kpi_summary
```

```python
# Sales by Region
sales_region = (
    df.groupby("Region")["Sales"]
      .sum()
      .reset_index()
      .sort_values(by="Sales", ascending=False)
)

sales_region
```

```python
# Profit by Region
profit_region = (
    df.groupby("Region")["Gross Profit"]
      .sum()
      .reset_index()
      .sort_values(by="Gross Profit", ascending=False)
)

profit_region
```

```python
# KPI Bar Chart
fig = px.bar(
    kpi_summary,
    x="KPI",
    y="Value",
    text="Value",
    title="Business KPIs"
)

fig.show()
```

```python
# Sales by Region Chart
fig = px.bar(
    sales_region,
    x="Region",
    y="Sales",
    text="Sales",
    title="Sales by Region"
)

fig.show()
```

```python
# Profit by Region Chart
fig = px.pie(
    profit_region,
    names="Region",
    values="Gross Profit",
    title="Profit Distribution by Region"
)

fig.show()
```

```python
# Top 10 Products by Sales
top_products = (
    df.groupby("Product Name")["Sales"]
      .sum()
      .reset_index()
      .sort_values(by="Sales", ascending=False)
      .head(10)
)

top_products
```

```python
# Product Sales Chart
fig = px.bar(
    top_products,
    x="Product Name",
    y="Sales",
    text="Sales",
    title="Top 10 Products by Sales"
)

fig.show()
```

```python
# Save Results
kpi_summary.to_csv(
    "../data/processed/kpi_summary.csv",
    index=False
)

sales_region.to_csv(
    "../data/processed/sales_by_region.csv",
    index=False
)

profit_region.to_csv(
    "../data/processed/profit_by_region.csv",
    index=False
)

top_products.to_csv(
    "../data/processed/top_products.csv",
    index=False
)
```
