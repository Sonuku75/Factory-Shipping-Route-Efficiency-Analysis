# Module 2: Data Cleaning and Validation

```python
import pandas as pd
import numpy as np
```

```python
# Load Dataset

df = pd.read_csv("../data/raw/Nassau Candy Distributor.csv")
```

```python
# Display Dataset

df.head()
```

```python
# Check data types

df.dtypes
```

```python
# Convert Order Date and Ship Date into datetime format

df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    dayfirst=True
)

df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    dayfirst=True
)
```

```python
df.dtypes
```

```python
# Check missing values

df.isnull().sum()
```

```python
# Check duplicate records

df.duplicated().sum()
```

```python
# Remove duplicate rows

df = df.drop_duplicates()
```

```python
# Calculate Shipping Lead Time

df["Shipping Lead Time"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days
```

```python
df[["Order Date","Ship Date","Shipping Lead Time"]].head()
```

```python
# Find invalid lead times

df[df["Shipping Lead Time"] < 0]
```

```python
# Remove negative lead times

df = df[df["Shipping Lead Time"] >= 0]
```

```python
df.shape
```

```python
# Save cleaned dataset

df.to_csv("../data/processed/cleaned_dataset.csv", index=False)
```

### Observation

- Date columns converted successfully.
- Shipping Lead Time calculated.
- Invalid lead times removed.
- Clean dataset saved for further analysis.
