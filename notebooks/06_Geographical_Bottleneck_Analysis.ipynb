# Module 6 : Geographical Bottleneck Analysis

```python
import pandas as pd
import plotly.express as px
```

```python
# Load Dataset

df = pd.read_csv("../data/processed/feature_engineered_dataset.csv")
```

```python
df.head()
```

```python
# Shipments by State
state_summary = (
    df.groupby("State/Province")
      .size()
      .reset_index(name="Total Shipments")
      .sort_values(
          by="Total Shipments",
          ascending=False
      )
)

state_summary.head(10)
```

```python
# Shipments by Region
region_summary = (
    df.groupby("Region")
      .size()
      .reset_index(name="Total Shipments")
      .sort_values(
          by="Total Shipments",
          ascending=False
      )
)

region_summary
```

```python
# Top 10 Bottleneck States
top10_states = state_summary.head(10)

top10_states
```

```python
# Top 10 Regions
top_regions = region_summary.head()

top_regions
```

```python
# State Bar Chart
fig = px.bar(
    top10_states,
    x="State/Province",
    y="Total Shipments",
    title="Top 10 States by Shipments",
    text="Total Shipments"
)

fig.show()
```

```python
# Region Pie Chart
fig = px.pie(
    region_summary,
    names="Region",
    values="Total Shipments",
    title="Shipment Distribution by Region"
)

fig.show()
```

```python
# State Wise Lead Time
state_leadtime = (
    df.groupby("State/Province")["Shipping Lead Time"]
      .mean()
      .reset_index(name="Average Lead Time")
      .sort_values(
          by="Average Lead Time",
          ascending=False
      )
)

state_leadtime.head(10)
```

```python
# Lead Time Chart
fig = px.bar(
    state_leadtime.head(10),
    x="State/Province",
    y="Average Lead Time",
    title="Top States by Average Lead Time",
    text="Average Lead Time"
)

fig.show()
```

```python
# Save Files
state_summary.to_csv(
    "../data/processed/state_summary.csv",
    index=False
)

region_summary.to_csv(
    "../data/processed/region_summary.csv",
    index=False
)

state_leadtime.to_csv(
    "../data/processed/state_leadtime.csv",
    index=False
)
```
