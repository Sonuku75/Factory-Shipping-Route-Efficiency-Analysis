# Module 7 : Ship Mode Performance Analysis

```python
import pandas as pd
import plotly.express as px
```

```python
# Load Dataset

df = pd.read_csv("../data/processed/feature_engineered_dataset.csv")
```

```python
df.head()
```

```python
# Ship Mode Summary
ship_mode_summary = (
    df.groupby("Ship Mode")
      .agg(
          Total_Shipments=("Ship Mode", "count"),
          Average_Lead_Time=("Shipping Lead Time", "mean"),
          Total_Sales=("Sales", "sum"),
          Total_Profit=("Gross Profit", "sum")
      )
      .reset_index()
)

ship_mode_summary
```

```python
# Most Used Ship Mode
most_used = ship_mode_summary.sort_values(
    by="Total_Shipments",
    ascending=False
)

most_used
```

```python
# Highest Sales Ship Mode
highest_sales = ship_mode_summary.sort_values(
    by="Total_Sales",
    ascending=False
)

highest_sales
```

```python
# Highest Profit Ship Mode
highest_profit = ship_mode_summary.sort_values(
    by="Total_Profit",
    ascending=False
)

highest_profit
```

```python
# Shipment Distribution Chart
fig = px.bar(
    ship_mode_summary,
    x="Ship Mode",
    y="Total_Shipments",
    text="Total_Shipments",
    title="Shipments by Ship Mode"
)

fig.show()
```

```python
# Sales Distribution Chart
fig = px.bar(
    ship_mode_summary,
    x="Ship Mode",
    y="Total_Sales",
    text="Total_Sales",
    title="Sales by Ship Mode"
)

fig.show()
```

```python
# Profit Distribution Chart
fig = px.pie(
    ship_mode_summary,
    names="Ship Mode",
    values="Total_Profit",
    title="Profit Distribution by Ship Mode"
)

fig.show()
```

```python
# Average Lead Time Chart
fig = px.bar(
    ship_mode_summary,
    x="Ship Mode",
    y="Average_Lead_Time",
    text="Average_Lead_Time",
    title="Average Lead Time by Ship Mode"
)

fig.show()
```

```python
# Save Results
ship_mode_summary.to_csv(
    "../data/processed/ship_mode_analysis.csv",
    index=False
)
```
