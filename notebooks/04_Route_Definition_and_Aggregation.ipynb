# Module 4: Route Definition and Aggregation

```python
import pandas as pd
import numpy as np
```

```python
# Load Feature Engineered Dataset

df = pd.read_csv("../data/processed/feature_engineered_dataset.csv")
```

```python
df.head()
```

```python
# Route-wise shipment count
route_count = (
    df.groupby("Factory_to_State")
      .size()
      .reset_index(name="Shipment Count")
)

route_count.head()
```

```python
# Average Lead Time
avg_lead_time = (
    df.groupby("Factory_to_State")["Shipping Lead Time"]
      .mean()
      .reset_index(name="Average Lead Time")
)

avg_lead_time.head()
```

```python
# Route Performence Summary
route_summary = (
    df.groupby("Factory_to_State")
      .agg(
          Shipment_Count=("Shipping Lead Time", "count"),
          Average_Lead_Time=("Shipping Lead Time", "mean"),
          Min_Lead_Time=("Shipping Lead Time", "min"),
          Max_Lead_Time=("Shipping Lead Time", "max")
      )
      .reset_index()
)

route_summary.head()
```

```python
# Ship Mode Count
ship_mode_summary = (
    df.groupby("Ship Mode")
      .size()
      .reset_index(name="Total Shipments")
)

ship_mode_summary
```

```python
# Average Lead Time by Ship Mode
ship_mode_performance = (
    df.groupby("Ship Mode")["Shipping Lead Time"]
      .mean()
      .reset_index(name="Average Lead Time")
)

ship_mode_performance
```

```python
# Save Results
route_summary.to_csv(
    "../data/processed/route_summary.csv",
    index=False
)

ship_mode_performance.to_csv(
    "../data/processed/ship_mode_performance.csv",
    index=False
)
```

### Observation

- Route-wise shipment statistics generated.
- Average lead time calculated.
- Ship mode performance analyzed.
- Summary files saved successfully.
